**Cell 1 — Imports and configuration**

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gumbel_r

OUTPUT_DIR = os.getcwd()

SEED = 123
N = 150
```

**Cell 2 — Simulate Gumbel sample**

```python
np.random.seed(SEED)
t = np.arange(1, N + 1) / (N + 1)

# R: rgumbel(n) = gumbel_r.rvs(loc=0, scale=1)
xf1 = gumbel_r.rvs(loc=0, scale=1, size=N, random_state=SEED)
xf = np.sort(xf1)
```

**Cell 3 — Compute PP-plot coordinates**

```python
emp = gumbel_r.cdf(xf)
```

**Cell 4 — Plot, display and save**

```python
fig, ax = plt.subplots(figsize=(5, 5))

ax.scatter(emp, t, color='blue', s=3)
ax.plot([0, 1], [0, 1], color='red', linewidth=0.5, zorder=2)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_xlabel("")
ax.set_ylabel("")
ax.grid(False)

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "pp_plot_Gumbel_CDF.png"),
            dpi=200, transparent=True, bbox_inches='tight')
plt.show()
```
