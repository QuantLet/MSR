```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

# Load data and extract second column
x = pd.read_csv("../Data/DE_portfolio_log_returns.csv", header=0)
xf = np.sort(x.iloc[:, 1].dropna().astype(float).values)

n = len(xf)
t = np.arange(1, n + 1) / (n + 1)

# PP Plot
theoretical_cdf = stats.norm.cdf(xf, loc=np.mean(xf), scale=np.std(xf))

fig1, ax1 = plt.subplots(figsize=(6, 5))
ax1.scatter(theoretical_cdf, t, color="blue", s=0.5)
ax1.plot([0, 1], [0, 1], color="red", linewidth=0.5)
plt.tight_layout()
plt.savefig("pp_plot.png", transparent=True, dpi=150)
plt.show(fig1)

# QQ Plot
(osm, osr), (slope, intercept, _) = stats.probplot(xf, dist="norm")

fig2, ax2 = plt.subplots(figsize=(6, 5))   # ← square figure
ax2.scatter(osm, osr, color="blue", s=0.5)
ax2.plot(osm, slope * np.array(osm) + intercept, color="red", linewidth=0.5)
ax2.set_ylim(-1, 1)
plt.tight_layout()
plt.savefig("qq_plot.png", transparent=True, dpi=150)
plt.show(fig2)
```
