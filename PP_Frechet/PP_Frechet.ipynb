**Cell 1 — Imports and configuration**

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import genextreme, norm

OUTPUT_DIR = os.getcwd()

SEED = 123
N = 100
ALPHA = 0.5
```

**Cell 2 — Simulate Fréchet sample**

```python
np.random.seed(SEED)
t = np.arange(1, N + 1) / (N + 1)

frec1 = genextreme.rvs(c=-ALPHA, loc=1, scale=ALPHA, size=N, random_state=SEED)
```

**Cell 3 — Compute PP-plot coordinates**

```python
frec = norm.cdf(np.sort(frec1))
```

**Cell 4 — Plot, display and save**

```python
fig, ax = plt.subplots(figsize=(5, 5))

ax.scatter(frec, t, color='blue', s=3)
ax.plot([0, 1], [0, 1], color='red', linewidth=0.5, zorder=2)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.grid(False)

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "pp_plot_Frechet.png"),
            dpi=200, transparent=True, bbox_inches='tight')
plt.show()
```
