```python
import yfinance as yf
import pandas as pd

# -----------------------------------------
# Download Bayer AG data from Yahoo Finance
# -----------------------------------------
ticker = "BAYN.DE"

data = yf.download(
    ticker,
    start="1900-01-01",   # Yahoo trims automatically
    progress=False
)

# Keep only Adjusted Close
data = data[["Close"]].dropna()

# Save to CSV
csv_file = "BAYN_DE_prices.csv"
data.to_csv(csv_file)

print(f"Data saved to {csv_file}")
print(data.head())
```

```python
import pandas as pd
import numpy as np

# -----------------------------------------
# Read data from CSV
# -----------------------------------------
prices = pd.read_csv(
    "BAYN_DE_prices.csv",
    index_col="Date",
    parse_dates=True
)["Close"]

# -----------------------------------------
# Compute log-returns
# -----------------------------------------
log_returns = np.log(prices / prices.shift(1)).dropna()

print("Log-returns preview:")
print(log_returns.head())
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

start_date = log_returns.index[0]
end_date = log_returns.index[-1]

fig, ax = plt.subplots(figsize=(10, 4))

ax.plot(log_returns.index, log_returns, color="blue", linewidth=0.5)

ax.set_xlabel("Date")
ax.set_ylabel("Log-return")

ax.set_xlim(start_date, end_date)

# ---- Fewer year labels (every 2 years) ----
ax.xaxis.set_major_locator(mdates.YearLocator(2))
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))

ax.grid(False)

fig.patch.set_alpha(0.0)
ax.patch.set_alpha(0.0)

plt.tight_layout()

plt.savefig(
    "BAYN_DE_log_returns.png",
    dpi=300,
    transparent=True,
    bbox_inches="tight"
)

plt.show()
```
