```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import genextreme, norm

np.random.seed(123)
n = 100
t = np.arange(1, n + 1) / (n + 1)

# Gumbel: rgev(n) → c=0, loc=0, scale=1
gumb1 = genextreme.rvs(c=0, loc=0, scale=1, size=n, random_state=123)
gumb  = norm.cdf(np.sort(gumb1))

fig, ax = plt.subplots(figsize=(5, 5))

ax.scatter(gumb, t, color='blue', s=5)
ax.plot([0, 1], [0, 1], color='red', linewidth=0.5, zorder=2)
# ax.set_title("PP Plot of Extreme Value - Gumbel", fontsize=9)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.grid(False)

plt.tight_layout()
plt.savefig("pp_plot_Gumbel.png", dpi=200, transparent=True, bbox_inches='tight')
plt.show()
```

**Cell 1 — Imports and configuration**

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import genextreme, norm

OUTPUT_DIR = os.getcwd()

SEED = 123
N = 100
```

**Cell 2 — Simulate Gumbel sample**

```python
np.random.seed(SEED)
t = np.arange(1, N + 1) / (N + 1)

# Gumbel: rgev(n) -> c=0, loc=0, scale=1
gumb1 = genextreme.rvs(c=0, loc=0, scale=1, size=N, random_state=SEED)
```

**Cell 3 — Compute Normal PP-plot coordinates**

```python
gumb = norm.cdf(np.sort(gumb1))
```

**Cell 4 — Plot, display and save**

```python
fig, ax = plt.subplots(figsize=(5, 5))

ax.scatter(gumb, t, color='blue', s=3)
ax.plot([0, 1], [0, 1], color='red', linewidth=0.5, zorder=2)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.grid(False)

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "pp_plot_Gumbel.png"),
            dpi=200, transparent=True, bbox_inches='tight')
plt.show()
```
